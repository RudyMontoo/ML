# Lecture 7 Autograd

```python
import torch
```

```python
material_cost=torch.tensor(10, requires_grad=True, dtype=torch.float16)
labor_cost=torch.tensor(5, requires_grad=True, dtype=torch.float16)

total_cost=3*material_cost**2+5*labor_cost+100
total_cost
```

```python
# .backward() function is used for automatic differentiation
total_cost.backward()
material_cost.grad, labor_cost.grad


# You are seeing this RuntimeError because the computation graph created by PyTorch for total_cost is freed 
# after the first backward pass. If you try to call .backward() a second time on the same variable without
# recreating the computation, PyTorch can't compute the gradients again. This is a memory optimization feature.
```
